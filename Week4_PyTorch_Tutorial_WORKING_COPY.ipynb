```python
# now, move both tensors to GPU
if not torch.cuda.is_available():
    raise RuntimeError("No GPU found. In Colab: Runtime > Change runtime type > T4 GPU")

if torch.cuda.is_available():
  device = torch.device("cuda")
  a = a.to(device)
  b = b.to(device)

print(f"a is on: {a.device}")
print(f"b is on: {b.device}")

start = time.perf_counter()
matrix_mult_gpu = a @ b
torch.cuda.synchronize()
gpu_time = time.perf_counter() - start

print(f"GPU matmul time: {gpu_time:.4f} s")
```

## Datasets and DataLoaders


* In PyTorch, a `Dataset` object store (X, y) pairs, and allows us to
* `DataLoader` objects alow us to batch and stream the Dataset to the model


A custom `Dataset` class needs three methods:

- `__init__`: store the data (here we convert it to tensors)
- `__len__`: return how many examples there are
- `__getitem__`: return one `(features, label)` pair by index


```python
# Implementing a custom Dataset
class XORDataset(Dataset):
    def __init__(self, data):
        # Runs once, when you create the dataset.
        self.data = data

    def __len__(self):
        # "How many examples are there?"
        return len(self.data)

    def __getitem__(self, idx):
        # Returns (input, target) at index idx as tensors.
        inputs, target = self.data[idx]
        x = torch.tensor(inputs, dtype=torch.float32)
        y = torch.tensor([target], dtype=torch.float32)
        return x, y

dataset = XORDataset(data)
```

## Train

Each step is the process from last class, with PyTorch doing the calculus:

1. **Forward pass**: compute predictions
2. **Loss**: compare predictions to the targets
3. **Backward pass**: `loss.backward()` computes the gradient for every weight. This is the backpropagation we did by hand.
4. **Update**: `optimizer.step()` adjusts the weights, then `optimizer.zero_grad()` resets the gradients before moving on to the next batch.

An **epoch** is one pass through the whole dataset.
